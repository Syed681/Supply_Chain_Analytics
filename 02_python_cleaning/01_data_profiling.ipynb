```python
# ============================================================
# 1. Import Libraries
# ============================================================
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
# ============================================================
# 2. LOAD DATA
# ============================================================
```

```python
df = pd.read_csv("DataCoSupplyChainDataset.csv",encoding="latin1")

print("\nDataset loaded successfully.")
```

```python
# ============================================================
# 3. BASIC DATASET INFORMATION
# ============================================================
```

```python
print("\n" + "=" * 70)
print("DATASET OVERVIEW")
print("=" * 70)

print(f"Rows       : {df.shape[0]:,}")
print(f"Columns    : {df.shape[1]:,}")
```

```python
# ============================================================
# 4. COLUMN INFORMATION
# ============================================================
```

```python
print("\n" + "=" * 70)
print("COLUMN INFORMATION")
print("=" * 70)

column_info = pd.DataFrame({
    "Column": df.columns,                     # Column names
    "Data_Type": df.dtypes.astype(str),       # Data type of each column
    "Null_Count": df.isnull().sum().values,   # Number of missing values per column
    "Null_Percentage": (
        df.isnull().mean().values * 100       # Percentage of missing values
    ).round(2),
    "Unique_Values": df.nunique().values      # Count of unique values per column
})


print(column_info.to_string(index=False))
```

```python
# ============================================================
# 5. DUPLICATES
# ============================================================
```

```python
print("\n" + "=" * 70)
print("DUPLICATE CHECK")
print("=" * 70)

duplicate_rows = df.duplicated().sum()

print(f"Duplicate rows: {duplicate_rows},")
```

```python
# ============================================================
# 6. MISSING VALUES
# ============================================================
```

```python
print("\n" + "=" * 70)
print("MISSING VALUES")
print("=" * 70)

missing = (
    df.isnull()
      .sum()
      .reset_index()
)
missing.columns = ["Column", "Missing_Count"]

missing["Missing_Percentage"]= (
    missing["Missing_Count"] / len(df) * 100 ).round(2)

missing = missing[
    missing["Missing_Count"] > 0
].sort_values(
    "Missing_Count",
    ascending=False
)

print(missing.to_string(index=False))

```

```python
# ============================================================
# 7. NUMERIC SUMMARY
# ============================================================
```

```python
print("\n" + "=" * 70)
print("NUMERIC DATA SUMMARY")
print("=" * 70)

desc = df.describe().T
desc["median"]=df.median(
    numeric_only=True )
print(desc.to_string())
```

```python
# ============================================================
# 8. CATEGORICAL SUMMARY
# ============================================================
```

```python
print("\n" + "=" * 70)
print("CATEGORICAL DATA SUMMARY")
print("=" * 70)

categorical_columns = df.select_dtypes(
    include=["object"]
).columns

for column in categorical_columns:

    print(f"\n--- {column} ---")

    print(
        df[column]
        .value_counts(dropna=False)
        .head(10)
        .to_string()
    )
```

```python
# ============================================================
# 9. KEY BUSINESS IDENTIFIERS
# ============================================================
```

```python
print("\n" + "=" * 70)
print("KEY BUSINESS IDENTIFIERS")
print("=" * 70)

key_columns = [
    "Order Item Id",
    "Order Id",
    "Customer Id",
    "Order Customer Id",
    "Product Card Id",
    "Order Item Cardprod Id",
    "Category Id",
    "Product Category Id",
    "Department Id"
]

for col in key_columns:
    if col in df.columns:   # prevents KeyError
        print(f"{col:<30} Unique: {df[col].nunique():,} | Rows: {len(df):,}")
    else:
        print(f"{col:<30} ❌ Not found in DataFrame")
```

```python
# ============================================================
# 10. DATE ANALYSIS
# ============================================================
```

```python
print("\n" + "=" * 70)
print("DATE ANALYSIS")
print("=" * 70)

date_columns = [
    "order date (DateOrders)",
    "shipping date (DateOrders)"
]

for col in date_columns:

    if col in df.columns:

        date_series = pd.to_datetime(
            df[col],
            errors="coerce"
        )

        print(f"\n{col}")
        print(f"Minimum : {date_series.min()}")
        print(f"Maximum : {date_series.max()}")
        print(f"Invalid : {date_series.isna().sum():,}")
```

```python
# ============================================================
# 11. SHIPPING PERFORMANCE
# ============================================================
```

```python
print("\n" + "=" * 70)
print("SHIPPING PERFORMANCE")
print("=" * 70)

if (
    "Days for shipping (real)" in df.columns
    and
    "Days for shipment (scheduled)" in df.columns
):

    df["Shipping_Variance"] = (
        df["Days for shipping (real)"]
        -
        df["Days for shipment (scheduled)"]
    )

    print(
        df["Shipping_Variance"]
        .describe()
        .to_string()
    )

    print("\nShipping variance distribution:")

    print(
        df["Shipping_Variance"]
        .value_counts()
        .sort_index()
        .to_string()
    )

```

```python
# ============================================================
# 12. DELIVERY STATUS
# ============================================================
```

```python
print("\n" + "=" * 70)
print("DELIVERY STATUS")
print("=" * 70)

if "Delivery Status" in df.columns:

    delivery_summary = (
        df["Delivery Status"]
        .value_counts()
        .to_frame("Order_Items")
    )

    delivery_summary["Percentage"] = (
        delivery_summary["Order_Items"]
        / len(df)
        * 100
    ).round(2)

    print(delivery_summary)

```

```python
# ============================================================
# 13. SHIPPING MODE
# ============================================================
```

```python
print("\n" + "=" * 70)
print("SHIPPING MODE")
print("=" * 70)

if "Shipping Mode" in df.columns:

    shipping_mode = (
        df["Shipping Mode"]
        .value_counts()
        .to_frame("Order_Items")
    )

    shipping_mode["Percentage"] = (
        shipping_mode["Order_Items"]
        / len(df)
        * 100
    ).round(2)

    print(shipping_mode)
```

```python
# ============================================================
# 14. PROFITABILITY CHECK
# ============================================================
```

```python
print("\n" + "=" * 70)
print("PROFITABILITY")
print("=" * 70)

if "Benefit per order" in df.columns:

    negative_profit = (
        df["Benefit per order"] < 0
    ).sum()

    print(
        f"Negative-profit order items: "
        f"{negative_profit:,}"
    )

    print(
        f"Negative-profit percentage: "
        f"{negative_profit / len(df) * 100:.2f}%"
    )


```

```python
# ============================================================
# 15. KEY BUSINESS METRICS
# ============================================================
```

```python
print("\n" + "=" * 70)
print("INITIAL BUSINESS METRICS")
print("=" * 70)

if "Order Id" in df.columns:
    print(
        f"Total Orders       : "
        f"{df['Order Id'].nunique():,}"
    )

if "Order Item Id" in df.columns:
    print(
        f"Total Order Items  : "
        f"{df['Order Item Id'].nunique():,}"
    )

if "Customer Id" in df.columns:
    print(
        f"Unique Customers   : "
        f"{df['Customer Id'].nunique():,}"
    )

if "Product Card Id" in df.columns:
    print(
        f"Unique Products    : "
        f"{df['Product Card Id'].nunique():,}"
    )

if "Sales" in df.columns:
    print(
        f"Total Sales        : "
        f"${df['Sales'].sum():,.2f}"
    )

if "Order Item Quantity" in df.columns:
    print(
        f"Total Quantity     : "
        f"{df['Order Item Quantity'].sum():,.0f}"
    )

if "Benefit per order" in df.columns:
    print(
        f"Total Profit       : "
        f"${df['Benefit per order'].sum():,.2f}"
    )


```
